# Volcano Eruptions

Which volcanoes have experienced the longest eruptions?

The dataset `volcanic-eruptions.csv` includes the start and end dates for all volcanic eruptions since 1800, along with each volcano's identification number. For eruptions that are still ongoing as of December 2024, the end date is recorded as December 2024.

This dataset excludes any eruptive pauses shorter than three months. If an eruption resumes after more than three months of inactivity, it is classified as a new eruption.

Note: Volcano Yasur in Vanuatu is not included in this list due to the absence of a clear start date.

```python
import pandas as pd
import matplotlib.pyplot as plt

eruptions = pd.read_csv('volcanic-eruptions.csv')
eruptions.head(3)
```

### Additional dataset

The dataset `volcano-list.csv` provides detailed information about each volcano, including its name, country, latitude, longitude, and type.

```python
volcanoes = pd.read_csv('volcano-list.csv')
volcanoes.head(3)
```

**Task 1: Find the volcanoes that were erupting as of Dec 2024.**

```python
eruptions.info()
```

Both the start and end date are in **string format**. Meaning, we have to covert the data to date time.

```python
eruptions['start_date'] = pd.to_datetime(eruptions['start_date'])
eruptions['end_date'] = pd.to_datetime(eruptions['end_date'])
eruptions.info()
```

```python
#getting the needed data from volcanoes dataset: volcano_id, volcano_name, country
volcano_info = volcanoes[['volcano_id','volcano_name', 'country']]
volcano_info
```

```python
#merging the data frames
volcano_analysis = eruptions.merge(volcano_info, on = 'volcano_id', how = 'left')
volcano_analysis
```

```python
# Finally, we can look for the volcanoes that erupted in 2024. 
volcano_2024 = volcano_analysis.query('end_date >= "2024-12-01"')
volcano_2024
```

**Task 2: Find the volcanoes that have had the longest volcanic eruptions.**

```python
#We can compute for the number of days each volcano erupted by subtracting the start date from the end date.
volcano_analysis['days'] = ((volcano_analysis['end_date'] - volcano_analysis['start_date']).dt.days)
volcano_analysis
```

```python
#We can sort the data by "days" and get the top 10 volcanoes that erupted for the longest time.
top_10_longest_eruptions = volcano_analysis.sort_values(by = 'days', ascending = False).head(10)
top_10_longest_eruptions
```

**ANSWER:** Based to the results, the Santa Maria Volcano in Guatemala erupted for the longest time, which lasted for 37439 days.
